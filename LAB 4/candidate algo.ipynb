```python
# Candidate Elimination Algorithm

data = [
    ['>=9', 'Yes', 'Excellent', 'Good', 'Fast', 'Yes', 'Yes'],
    ['>=9', 'Yes', 'Good', 'Good', 'Fast', 'Yes', 'Yes'],
    ['>=8', 'No', 'Good', 'Good', 'Fast', 'No', 'No'],
    ['>=9', 'Yes', 'Good', 'Good', 'Slow', 'No', 'Yes']
]

print("******** TRAINING DATA ********")
for row in data:
    print(row)

# Initialize S and G
S = data[0][:-1]
G = [['?' for i in range(len(S))]]

print("\nInitial Specific Hypothesis (S):")
print(S)

print("\nInitial General Hypothesis (G):")
for g in G:
    print(g)

step = 1

for row in data:

    print("\n===================================")
    print("Training Example", step)
    print(row)

    if row[-1] == "Yes":
        print("Positive Example")

        for i in range(len(S)):
            if S[i] != row[i]:
                S[i] = '?'

    else:
        print("Negative Example")

        new_G = []

        for i in range(len(S)):
            if S[i] != '?' and S[i] != row[i]:
                temp = ['?' for j in range(len(S))]
                temp[i] = S[i]
                new_G.append(temp)

        G = new_G

    print("\nSpecific Hypothesis (S):")
    print(S)

    print("\nGeneral Hypothesis (G):")
    for g in G:
        print(g)

    step += 1

print("\n***********************************")
print("Final Specific Hypothesis (S):")
print(S)

print("\nFinal General Hypothesis (G):")
for g in G:
    print(g)
```
