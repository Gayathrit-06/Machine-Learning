```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create the dataset
data = {
    "Week": [1, 2, 3, 4, 5],
    "Sales": [2, 3, 4, 5, 6]
}

# Create DataFrame
df = pd.DataFrame(data)

# Select input (X) and output (Y)
x = df[["Week"]]
y = df["Sales"]

# Create and train the model
model = LinearRegression()
model.fit(x, y)

# Predict sales
predictions = model.predict(x)

# Print results
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"Sales = {model.coef_[0]:.2f} * Week + {model.intercept_:.2f}")

# R² Score
print(f"\nR² Score: {model.score(x, y):.4f}")

# Plot data
plt.figure(figsize=(6,4))
plt.scatter(x, y, color="green", label="Actual Data")
plt.plot(x, predictions, color="blue", linewidth=2, label="Regression Line")

plt.title("Linear Regression: Week vs Sales")
plt.xlabel("Week")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

# Generate random dataset
x, y = make_regression(
    n_samples=50,
    n_features=1,
    noise=10,
    random_state=42
)

# Plot dataset
plt.figure(figsize=(6,4))
plt.scatter(x, y, color="green", label="Data Points")

plt.title("Linear Regression using Random Dataset")
plt.xlabel("X")
plt.ylabel("Y")

# Train model
model = LinearRegression()
model.fit(x, y)

# Predict values
predictions = model.predict(x)

# Print model parameters
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"y = {model.coef_[0]:.2f} * x + {model.intercept_:.2f}")

# R² Score
print(f"\nR² Score: {model.score(x, y):.4f}")

# Plot regression line
plt.plot(x, predictions, color="red", linewidth=2, label="Regression Line")

plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

print("Multiple Linear Regression")

# Generate dataset
x, y = make_regression(
    n_samples=50,
    n_features=3,
    noise=0.1,
    random_state=42
)

# Create model
model = LinearRegression()

# Train model
model.fit(x, y)

# Print intercept
print("Intercept:", model.intercept_)

# Print coefficients
print("\nCoefficients:")
for i, coef in enumerate(model.coef_, start=1):
    print(f"x{i}: {coef:.2f}")

# Regression equation
print("\nRegression Equation:")
print(
    f"y = {model.coef_[0]:.2f} * x1 + "
    f"{model.coef_[1]:.2f} * x2 + "
    f"{model.coef_[2]:.2f} * x3 + "
    f"{model.intercept_:.2f}"
)

# R² Score
print(f"\nR² Score: {model.score(x, y):.4f}")
```

```python
import numpy as np
from sklearn.decomposition import PCA

# Original dataset
x = np.array([
    [2, 6],
    [1, 7]
])

print("Original Matrix:")
print(x)

print("Shape:", x.shape)

# Create PCA model
pca = PCA(n_components=2)

# Apply PCA
x_pca = pca.fit_transform(x)

print("\nTransformed Matrix:")
print(x_pca)

print("Shape:", x_pca.shape)

# Inverse transform
x_reconstructed = pca.inverse_transform(x_pca)

print("\nReconstructed Matrix:")
print(x_reconstructed)

# Explained variance ratio
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Singular values
print("\nSingular Values:")
print(pca.singular_values_)

# Principal components
print("\nPrincipal Components:")
print(pca.components_)

# Mean
print("\nMean:")
print(pca.mean_)
```

```python
import numpy as np

# Define matrix
A = np.array([
    [1, 2],
    [3, 1],
    [5, 6]
])

print("Original Matrix (A):")
print(A)

# Perform SVD
U, S, VT = np.linalg.svd(A)

print("\nMatrix U:")
print(U)

print("\nSingular Values:")
print(S)

print("\nMatrix Vᵀ:")
print(VT)

# Create Sigma matrix
Sigma = np.zeros((A.shape[0], A.shape[1]))
np.fill_diagonal(Sigma, S)

print("\nSigma Matrix:")
print(Sigma)

# Reconstruct matrix
A_reconstructed = U @ Sigma @ VT

print("\nReconstructed Matrix:")
print(A_reconstructed)

# Verification
print("\nVerification:", np.allclose(A, A_reconstructed))
```
