```python

states = ('Good', 'Moderate', 'Fail')
observation = ('Happy', 'Worry', 'Blink', 'Fine')

initial_probability = {
    'Good': 0.5,
    'Moderate': 0.25,
    'Fail': 0.25
}

transition_probability = {
    'Good': {'Good': 0.4, 'Moderate': 0.3, 'Fail': 0.3},
    'Moderate': {'Good': 0.6, 'Moderate': 0.25, 'Fail': 0.15},
    'Fail': {'Good': 0.3, 'Moderate': 0.4, 'Fail': 0.3}
}

observation_probability = {
    'Good': {'Happy': 0.4, 'Worry': 0.1, 'Blink': 0.2, 'Fine': 0.3},
    'Moderate': {'Happy': 0.3, 'Worry': 0.2, 'Blink': 0.3, 'Fine': 0.2},
    'Fail': {'Happy': 0.1, 'Worry': 0.4, 'Blink': 0.3, 'Fine': 0.2}
}


def viterbi(obs, states, i_pro, t_pro, o_pro):
    path = {s: [] for s in states}
    alpha_pro = {}

    # Initialization
    for s in states:
        alpha_pro[s] = i_pro[s] * o_pro[s][obs[0]]

    print("Initialization:")
    print(alpha_pro)

    # Recursion
    for i in range(1, len(obs)):
        last_pro = alpha_pro
        alpha_pro = {}
        new_path = {}

        for curr_state in states:
            max_pro, last_state = max(
                (
                    last_pro[prev_state]
                    * t_pro[prev_state][curr_state]
                    * o_pro[curr_state][obs[i]],
                    prev_state
                )
                for prev_state in states
            )

            alpha_pro[curr_state] = max_pro
            new_path[curr_state] = path[last_state] + [last_state]

        path = new_path
        print("Step", i + 1, ":", alpha_pro)

    last_state = max(states, key=lambda s: alpha_pro[s])
    max_path = path[last_state] + [last_state]

    return max_path


if __name__ == '__main__':
    obs = ['Happy', 'Blink', 'Fine', 'Fine']

    print("Given Observation sequence:")
    print(obs)

    result = viterbi(
        obs,
        states,
        initial_probability,
        transition_probability,
        observation_probability
    )

    print("Maximum Probability Path:")
    print(result)
```
