```python
# Find-S Algorithm

data = [
    ['>=9', 'Yes', 'Excellent', 'Good', 'Fast', 'Yes', 'Yes'],
    ['>=9', 'Yes', 'Good', 'Good', 'Fast', 'Yes', 'Yes'],
    ['>=8', 'No', 'Good', 'Good', 'Fast', 'No', 'No'],
    ['>=9', 'Yes', 'Good', 'Good', 'Slow', 'No', 'Yes']
]

print("******** TRAINING DATA ********")
for row in data:
    print(row)


hypothesis = data[0][:-1]

print("\nInitial Hypothesis:")
print(hypothesis)

step = 1

for row in data[1:]:
    if row[-1] == "Yes":
        for i in range(len(hypothesis)):
            if hypothesis[i] != row[i]:
                hypothesis[i] = '?'

        print("\nHypothesis after Positive Example", step)
        print(hypothesis)
        step += 1

print("\n******** FINAL HYPOTHESIS ********")
print(hypothesis)
```
