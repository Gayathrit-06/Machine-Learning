```python
# ID3 - based Decision Tree

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder

data = {
    'CGPA': ['9','8','9','10','9','9','10','9','8','9'],
    'Intern': ['Y','N','N','N','Y','Y','N','Y','Y','Y'],
    'PK': ['+++','+','==','==','+','+','+','+++','+','+'],
    'CG': ['G','M','P','G','M','M','M','G','G','G'],
    'Job': ['Y','Y','N','N','Y','Y','N','Y','Y','Y']
}

table = pd.DataFrame(data, columns=['CGPA', 'Intern', 'PK', 'CG', 'Job'])

table = table.where(table['CGPA'] == '9').dropna()

encoder = LabelEncoder()

for column in table.columns:
    table[column] = encoder.fit_transform(table[column])

X = table.iloc[:, 0:4].values
y = table.iloc[:, 4].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=2
)

model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=3
)

model = model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", metrics.accuracy_score(y_test, y_pred))

print(confusion_matrix(y_test, y_pred))

print(classification_report(y_test, y_pred))

print([1, 0, 0, 1])

if model.predict([[1, 0, 0, 1]]) == 1:
    print("Got Job")
else:
    print("Didn't get Job")

print([2, 0, 2, 0])
```

```python
# ID3 - Based Decision Tree

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder

data = {
    'CGPA': ['9','8','9','10','9','9','10','9','8','9'],
    'Intern': ['Y','N','N','N','Y','Y','N','Y','Y','Y'],
    'PK': ['+++','+','==','==','+','+','+','+++','+','+'],
    'CG': ['G','M','P','G','M','M','G','G','G','G'],
    'Job': ['Y','Y','N','N','Y','Y','N','Y','Y','Y']
}

table = pd.DataFrame(data)

# Encoding
for column in table.columns:
    encoder = LabelEncoder()
    table[column] = encoder.fit_transform(table[column])

X = table.iloc[:, 0:4]
y = table.iloc[:, 4]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=2, stratify=y
)

model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=3,
    random_state=2
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", metrics.accuracy_score(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("Classification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

student1 = pd.DataFrame(
    [[2, 0, 2, 0]],
    columns=X.columns
)

print("Student 1:", student1.values.tolist())

if model.predict(student1)[0] == 1:
    print("Got Job")
else:
    print("Didn't get Job")

student2 = pd.DataFrame(
    [[1, 0, 1, 1]],
    columns=X.columns
)

print("Student 2:", student2.values.tolist())

if model.predict(student2)[0] == 1:
    print("Got Job")
else:
    print("Didn't get Job")
```
